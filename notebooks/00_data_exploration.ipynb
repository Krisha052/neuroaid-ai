# 00 - Data Exploration

Goal: explore the real sample audio/transcripts used elsewhere in this project, sanity-check prompt coverage, and look at basic audio stats.

No real dyslexia-labeled dataset is used anywhere in this project -- see `data/sample_audio/real_samples/README.md` and `data/generated/README.md` for why, and how training labels are generated synthetically instead.

```python
import json
import sys
from pathlib import Path

sys.path.insert(0, '..')
from src.audio.diagnostics import wav_info

REAL_AUDIO_DIR = Path('../data/sample_audio/real_samples')
transcripts = json.loads((REAL_AUDIO_DIR / 'transcripts.json').read_text())
transcripts
```

```python
for clip_id in transcripts:
    info = wav_info(REAL_AUDIO_DIR / f'{clip_id}.wav')
    print(clip_id, info)
```

Each clip is a few seconds of real (LibriSpeech, CC BY 4.0) read-speech audio at 16kHz mono -- the format the rest of the pipeline expects. `scripts/generate_training_data.py` extracts real acoustic features (MFCC, pause ratio) from these same clips.
